Given an array of integers nums, sort the array in ascending order and return it.

You must solve the problem without using any built-in functions in O(nlog(n)) time complexity and with the smallest space complexity possible.



Example 1:

Input: nums = [5,2,3,1]
Output: [1,2,3,5]
Explanation: After sorting the array, the positions of some numbers are not changed (for example, 2 and 3), while the positions of other numbers are changed (for example, 1 and 5).

Example 2:

Input: nums = [5,1,1,2,0,0]
Output: [0,0,1,1,2,5]
Explanation: Note that the values of nums are not necessarily unique.



Constraints:

    1 <= nums.length <= 5 * 104
    -5 * 104 <= nums[i] <= 5 * 104



# Selection Sort

```python
from typing import List
```

```python
class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        if not nums:
            return []
        i=0
        while i < len(nums):
            smallest = float('inf')
            for index in range(i, len(nums)):
                if nums[index] < smallest:
                    smallest = nums[index]
                    smallest_index = index
            nums[i], nums[smallest_index] = nums[smallest_index], nums[i]
            i += 1
        return nums
```

```python
dsa = Solution()
```

```python
dsa.sortArray(nums=[3,2,1, 1, 2, 4, 85])
```

```python
# Cleaner Code

class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        n = len(nums)

        for i in range(n):
            min_index = i

            for j in range(i + 1, n):
                if nums[j] < nums[min_index]:
                    min_index = j

            nums[i], nums[min_index] = nums[min_index], nums[i]

        return nums
```

# Bubble Sort

```python
class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        i = len(nums) - 1

        while i != 0:
          swapped = False
          for j in range(i):
            if nums[j] > nums[j+1]:
              nums[j], nums[j+1] = nums[j+1], nums[j]
              swapped = True
          i -= 1
          if not swapped:
            break
        return nums
```

| Feature                       | Selection Sort                                      | Bubble Sort                                          |
| ----------------------------- | --------------------------------------------------- | ---------------------------------------------------- |
| **Core idea**                 | Select the minimum and place it in correct position | Repeatedly swap adjacent elements to push max to end |
| **How it works**              | Finds smallest from unsorted part and swaps once    | Compares adjacent pairs and swaps many times         |
| **Swaps**                     | 🔽 Minimum (at most `n-1`)                          | 🔼 Many swaps                                        |
| **Comparisons**               | Always `O(n²)`                                      | Always `O(n²)`                                       |
| **Best case**                 | `O(n²)` ❌ (no improvement)                          | `O(n)` ✅ (with optimization)                         |
| **Worst case**                | `O(n²)`                                             | `O(n²)`                                              |
| **Adaptive?**                 | ❌ No                                                | ✅ Yes (with swapped flag)                            |
| **Stable?**                   | ❌ Not stable                                        | ✅ Stable                                             |
| **When array already sorted** | Still scans everything                              | Stops early (optimized version)                      |
| **Use in real life**          | Rare                                                | Rare (but slightly more useful than selection)       |


🧠 What does “stable sorting” mean?

A sorting algorithm is stable if:

It preserves the relative order of equal elements

📌 Example to understand

Let’s say you have:

[(2, 'A'), (1, 'B'), (2, 'C')]

You’re sorting based on the number only.

✅ Stable sort result
[(1, 'B'), (2, 'A'), (2, 'C')]

Notice:

(2, 'A') comes before (2, 'C')
👉 Same order as original
❌ Unstable sort result
[(1, 'B'), (2, 'C'), (2, 'A')]

Now:

Order of equal elements changed ❌

# Insertion Sort

```python
# Cleaner

class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        for i in range(1, len(nums)):
            key = nums[i]
            j = i - 1

            while j >= 0 and nums[j] > key:
                nums[j + 1], nums[j] = nums[j], nums[j + 1]
                j -= 1

        return nums
```

| Feature                      | Selection Sort                       | Bubble Sort                      | Insertion Sort                       |
| ---------------------------- | ------------------------------------ | -------------------------------- | ------------------------------------ |
| **Core idea**                | Pick minimum and place it            | Swap adjacent to push max to end | Insert element into correct position |
| **How it works**             | Find min in unsorted part, swap once | Repeated adjacent swaps          | Shift elements and insert            |
| **Swaps / Moves**            | 🔽 Very few swaps                    | 🔼 Many swaps                    | 🔁 Mostly shifts (efficient)         |
| **Best case**                | O(n²) ❌                              | O(n) ✅ (optimized)               | O(n) ✅                               |
| **Worst case**               | O(n²)                                | O(n²)                            | O(n²)                                |
| **Adaptive?**                | ❌ No                                 | ✅ Yes                            | ✅ Yes                                |
| **Stable?**                  | ❌ No                                 | ✅ Yes                            | ✅ Yes                                |
| **When array nearly sorted** | Still slow ❌                         | Faster                           | 🚀 Very fast                         |
| **Real-world usage**         | Rare                                 | Rare                             | ✅ Used (small / nearly sorted data)  |
| **Conceptual style**         | Selection                            | Repeated correction              | Incremental building                 |


🧠 Intuition summary
Selection Sort
👉 “Find the correct element and place it”
Bubble Sort
👉 “Fix local mistakes repeatedly”
Insertion Sort
👉 “Build sorted list by inserting”

```python
class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
      print(f'nums: {nums}')
      if not nums:
        return []
      elif len(nums) == 1:
        return nums

      def merge(left, right):
        start_left, start_right = 0, 0
        arr = []
        print(f"left: {left}")
        print(f"right: {right}")
        while start_left < len(left) and start_right < len(right):
          if left[start_left] < right[start_right]:
            arr.append(left[start_left])
            start_left += 1
          else:
            arr.append(right[start_right])
            start_right += 1
        arr.extend(left[start_left:])
        arr.extend(right[start_right:])
        print(f"arr: {arr}")
        return arr

      mid = len(nums)//2
      left = nums[:mid]
      right = nums[mid:]
      left = self.sortArray(left)
      right = self.sortArray(right)
      arr = merge(left, right)

      return arr
```

```python
dsa = Solution()
```

```python
dsa.sortArray(nums=[3,2,1, 4,7, 3])
```

Yes — exactly. Using indices instead of slicing is the key optimization.

Let’s make it intuitive.

🔴 Why slicing is inefficient
left = nums[:mid]
right = nums[mid:]

Every time you do this:

You create new arrays
Copy elements → O(n) extra work per level
Across recursion → O(n log n) extra space
🟢 Using indices (optimized approach)

Instead of creating new lists, you:

Pass start and end indices
Work on the same original array
Use one helper array

```python
arr = [4, 6, 2, 5, 7, 9, 1, 3]
```

```python
def quick_sort(array, left, right):

    if left >= right:
        return array

    pivot = array[left]
    i = left + 1
    j = right

    while True:

        while i <= j and array[i] <= pivot:
            i += 1

        while i <= j and array[j] >= pivot:
            j -= 1

        if i > j:
            break

        array[i], array[j] = array[j], array[i]

    # Put pivot in its correct position
    array[left], array[j] = array[j], array[left]
    print(array)
    quick_sort(array, left, j - 1)
    quick_sort(array, j + 1, right)

    return array
```
